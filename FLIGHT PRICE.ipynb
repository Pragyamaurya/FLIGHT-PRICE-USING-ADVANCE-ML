**4. Convert Duration Column into Numeric Form "Duration" is in text format (e.g., "2h 50m"). We need to convert it into total minutes.**

```python
# Convert Duration to total minutes
def convert_duration(duration):
    h, m = 0, 0
    if "h" in duration:
        h = int(duration.split("h")[0])
    if "m" in duration:
        m = int(duration.split("m")[0].split()[-1])
    return h * 60 + m

data["Duration_Minutes"] = data["Duration"].apply(convert_duration)

# Drop original Duration column
data.drop(columns=["Duration"], inplace=True)
```
